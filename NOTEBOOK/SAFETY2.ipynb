```python

import pandas as pd
from pathlib import Path

# Project directory
PROJECT_DIR = Path.cwd().parent

# Existing processed dataset directory
PROCESSED_DIR = PROJECT_DIR / "DATA" / "processed"

# Locate the already-extracted 2023 dataset
csv_files = [
    f for f in PROCESSED_DIR.glob("*.csv")
    if "ITA Case Detail Data 2023" in f.name
    and "checkpoint" not in f.name.lower()
]

print("Project directory:", PROJECT_DIR)
print("Processed directory exists:", PROCESSED_DIR.exists())
print("Matching datasets:", len(csv_files))

for f in csv_files:
    print(f.name)

if not csv_files:
    raise FileNotFoundError(
        "The 2023 ITA dataset was not found in DATA/processed."
    )

# Load the existing dataset
dataset_path = csv_files[0]
df = pd.read_csv(dataset_path, encoding="latin1", low_memory=False)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nAvailable columns:")
print(df.columns.tolist())
```

```python

# Inspect industry-related columns
industry_columns = [
    col for col in df.columns
    if any(keyword in col.lower()
           for keyword in ["naics", "industry", "company"])
]

print("Industry-related columns:")
print(industry_columns)

# Inspect industry descriptions and NAICS codes
for col in ["naics_code", "industry_description", "industry_type"]:
    if col in df.columns:
        print(f"\nTop values in {col}:")
        print(df[col].value_counts(dropna=False).head(20))
```

```python

# Search industry descriptions for petrochemical-related industries
keywords = [
    "petrochemical",
    "petroleum refinery",
    "petroleum refining",
    "refinery",
    "basic chemical",
    "organic chemical",
    "industrial chemical",
    "plastic resin",
    "synthetic rubber",
    "chemical manufacturing",
    "chemical plant"
]

# Find matching industry descriptions
industry_desc = df["industry_description"].fillna("").astype(str)

mask = industry_desc.str.contains(
    "|".join(keywords),
    case=False,
    regex=True
)

print("Petrochemical-related industry descriptions:")
print(
    df.loc[mask, "industry_description"]
      .value_counts()
      .to_string()
)

print("\nTotal matching incident records:", int(mask.sum()))

# Inspect relevant NAICS codes
naics = df["naics_code"].astype("string").str.strip()

naics_mask = naics.str.startswith(("325110", "324110"), na=False)

print("\nRecords with NAICS 325110 or 324110:")
print(int(naics_mask.sum()))

print("\nMatching NAICS code counts:")
print(df.loc[naics_mask, "naics_code"].value_counts().to_string())
```
